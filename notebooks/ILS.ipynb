```python
import pandas as pd

from cvrptw.benchmark import run_benchmark
from cvrptw.bks import compare_to_bks
from cvrptw.viz import draw_best_solutions, plot_ils_stats
```

## Benchmark

```python
results = run_benchmark(perturbation_moves=5)
```

## Results

```python
res_df = pd.DataFrame([vars(r) for r in results]).rename(
    columns={'n_iters': 'done_iters', 'elapsed': 'time'}
)
res_df
```

## Comparison with best-known solutions

Best-known values from [SINTEF TOP](https://www.sintef.no/projectweb/top/vrptw/solomon-benchmark/100-customers/). Both use the hierarchical objective: minimize vehicles first, then total distance (`ils(minimize_vehicles=False)` restores distance-only).

```python
bks_df = pd.DataFrame([vars(r) for r in compare_to_bks(results)])
print(f'mean gap: {bks_df.gap_pct.mean():+.2f}%, '
      f'at/below BKS distance: {(bks_df.distance <= bks_df.bks_distance).sum()}/{len(bks_df)}')
bks_df.sort_values('gap_pct', ascending=False)
```

```python
pd.DataFrame([r.operator_totals for r in results], index=[r.name for r in results])
```

```python
draw_best_solutions(res_df[['name', 'distance', 'n_vehicles', 'solution']].values)
```

## ILS Stats

```python
# Per-iteration stats for a single instance (change index to inspect others)
r = results[0]
stats_df = pd.DataFrame([vars(s) for s in r.stats])
stats_df.insert(0, 'iter', range(1, len(stats_df) + 1))
stats_df
```

```python
plot_ils_stats(r.stats, title=r.name)
```
