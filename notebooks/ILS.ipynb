```python
import pandas as pd

from cvrptw.benchmark import run_benchmark
from cvrptw.viz import draw_best_solutions, plot_ils_stats
```

## Benchmark

```python
results = run_benchmark(perturbation_moves=5)
```

## Results

```python
res_df = pd.DataFrame([vars(r) for r in results]).rename(
    columns={'n_iters': 'done_iters', 'elapsed': 'time'}
)
res_df
```

```python
draw_best_solutions(res_df[['name', 'distance', 'n_vehicles', 'solution']].values)
```

## ILS Stats

```python
# Per-iteration stats for a single instance (change index to inspect others)
r = results[0]
stats_df = pd.DataFrame([vars(s) for s in r.stats])
stats_df.insert(0, 'iter', range(1, len(stats_df) + 1))
stats_df
```

```python
plot_ils_stats(r.stats, title=r.name)
```
